## Функции

```python
def load_trajectories(instance_id, parquet_dir=PARQUET_DIR):
    dataset = ds.dataset(parquet_dir, format="parquet")
    table = dataset.to_table(filter=ds.field("instance_id") == instance_id)
    if len(table) == 0:
        raise ValueError(f"Не найден: {instance_id}")
    d = table.to_pydict()
    return d["trajectory"], d.get("exit_status", [])
```

```python
def show_step(traj, step_idx, context=1):
    for i in range(max(0, step_idx - context), step_idx):
        print(f"[{i}] {traj[i].get('role', '?')}: {traj[i].get('text', '')[:2000]}...")
    print("-" * 50)
    print(f"[{step_idx}] {traj[step_idx].get('role', '?')}:")
    print(traj[step_idx].get("text", ""))
    if step_idx < len(traj) - 1:
        print("-" * 50)
        print(f"[{step_idx+1}] {traj[step_idx+1].get('role', '?')}: {traj[step_idx+1].get('text', '')[:2000]}...")
```

```python
def find_errors_in_trajectory(instance_id, traj_idx, df=errors_df):
    """Найти все ошибки в конкретной траектории."""
    row = df[df["instance_id"] == instance_id]
    result = []
    for _, r in row.iterrows():
        for loc in r["locations"]:
            if loc["traj_idx"] == traj_idx:
                result.append((r["category"], loc["step_idx"], r["count"]))
    return result
```

```python
def list_trajectories(instance_id, parquet_dir=PARQUET_DIR):
    """Список траекторий с метаданными."""
    trajs, exits = load_trajectories(instance_id, parquet_dir)
    rows = []
    for i, (traj, exit_s) in enumerate(zip(trajs, exits)):
        rows.append({"idx": i, "steps": len(traj), "exit": exit_s})
    return pd.DataFrame(rows)
```
